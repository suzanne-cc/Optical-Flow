## Problem 1 - Loading and Displaying a Toy Problem

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from skimage import io, color
from IPython.display import HTML
import glob
```

```python
files = sorted(glob.glob("toy_problem/*.png"))
frames = []

for f in files:
    img = io.imread(f) # load
    img = color.rgb2gray(img) # convert to grayscale
    frames.append(img)

V = np.stack(frames, axis=2) # (256, 256, 64)

# displaying frames (using FuncAnimation)
fig, ax = plt.subplots()
img = ax.imshow(V[:, :, 0], cmap='gray')

def update(t):
    img.set_array(V[:, :, t])
    return [img]

animation = FuncAnimation(fig, update, frames=V.shape[2], interval=50)
plt.close(fig)

display(HTML(animation.to_jshtml()))
```
